```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Jawaban 2: Kolom menu pada baris ke-4, dan kolom terjual pada baris ke-2. Akan terjadi kesalahan perhitungan dan muncul error pada program disebabkan adanya nilai yang kosong.

```python
print(df.columns)
```

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Jawaban 4: kolom 'terjual' diisi fillna artinya tidak ada transaksi pada hari itu. Mengubah nilai kosong menjadi angka 0 secara logis mempertahankan data baris tersebut tanpa merusak perhitungan statistik (seperti total atau rata-rata penjualan).
kolom 'menu' diisi dropna guna untuk menghapus kolom yang kosong. karena tidak ada informasinya. Jadi menghapus kolom tersebut merupakan langkah yang baik.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Jawaban 5: Jumlah baris dataset sekarang akan berkurang jika sebelumnya terdapat data duplikat, atau tetap sama jika tidak ada baris yang duplikat. Untuk mencegah error kalkulasi dan akurasi analisis.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

```python
import pandas as pd
import numpy as np

# --- BAGIAN PANDAS ---
# Membaca dataset langsung dari file CSV (tidak perlu ketik manual lagi)
nama_file = "dataset_sensor_iot_kelas - dataset_sensor_iot_kelas.csv"
df = pd.read_csv(nama_file)

print("=== Tampilan Data Sensor ===")
# Menampilkan 5 baris pertama saja agar rapi
print(df.head())
print("-" * 40)

# --- BAGIAN NUMPY ---
# Mengambil kolom 'kelembapan_persen' dan mengubahnya jadi array NumPy
# Catatan: .dropna() dipakai untuk membuang baris yang datanya kosong (NaN/None)
kelembapan = np.array(df['kelembapan_persen'].dropna())

print("\n=== Perhitungan Kelembapan ===")
print('Rata-rata kelembapan udara:', kelembapan.mean())
print('Kelembapan udara tertinggi:', kelembapan.max())

# Mengibaratkan diskon 10% sebagai kalibrasi/pengurangan nilai sensor sebesar 10%
print('Kelembapan setelah dikalibrasi (dikurangi 10%):')
print(kelembapan * 0.9)
```
